# Convert Acoustic Data to Neurological

```python
def convert_acoustic_to_neurological(
    IXD: np.ndarray,
    Az: np.ndarray,
    El: np.ndarray,
    f: np.ndarray,
    target: str = "Az",
    tol: float = 1.5,
    verbose: int = 0,
) -> np.ndarray:
    """Convert acoustic profiles into averaged neurological profiles.

    Returns
    -------
    np.ndarray
        Array with shape:
        (directions, frequencies, target_axis_values)
    """
    IXD = np.asarray(IXD)
    Az = np.asarray(Az)
    El = np.asarray(El)
    f = np.asarray(f)

    if IXD.ndim != 2:
        raise ValueError("IXD must have shape (directions, frequencies).")

    direction_count, frequency_count = IXD.shape

    if Az.shape != (direction_count,) or El.shape != (direction_count,):
        raise ValueError("Az and El must match IXD's direction dimension.")

    if f.shape != (frequency_count,):
        raise ValueError("f must match IXD's frequency dimension.")

    if target not in {"Az", "El"}:
        raise ValueError("target must be either 'Az' or 'El'.")

    target_coordinates = Az if target == "Az" else El
    _, target_group = np.unique(
        target_coordinates,
        return_inverse=True,
    )
    target_count = target_group.max() + 1

    # Shape: source_direction, candidate_direction, frequency
    matches = np.isclose(
        IXD[:, None, :],
        IXD[None, :, :],
        atol=tol,
        rtol=0,
    )

    result = np.empty(
        (direction_count, frequency_count, target_count),
        dtype=np.float32,
    )

    # Soft version - Gaussian similarity
    sigma = tol / 6.0
    similarity = np.exp(
        -0.5 * ((IXD[:, None, :] - IXD[None, :, :]) / sigma) ** 2
    ) #** 4 # raise to power to sharpen the similarity function

    # # Kill weak activations
    # similarity = np.where(similarity > 0.6, similarity, 0.0)

    for group_index in range(target_count):
        group_mask = target_group == group_index
        #result[:, :, group_index] = matches[:, group_mask, :].mean(axis=1)
        result[:, :, group_index] = similarity[:, group_mask, :].max(axis=1)

    if verbose:
        print(f"Input shape: {IXD.shape}")
        print(f"Output shape: {result.shape}")
        print(f"Target groups: {target_count}")

    return result


def plot_neurological_profile(
    profile: np.ndarray,
    aspect: str = "auto",
    origin: str = "lower",
    target: np.ndarray = None,
    mode: str = "Azimuth",
    f: np.ndarray = None,
    ax=None,
):
    """Plot one neurological profile on an optional Matplotlib axis."""
    ax = plt.gca() if ax is None else ax
    image = ax.imshow(
        profile,
        aspect=aspect,
        origin=origin,
        extent=[target.min(), target.max(), f.min(), f.max()],
    )
    ax.set_xlabel(f"{mode} (deg)")
    ax.set_ylabel("Frequency")
    ax.figure.colorbar(image, ax=ax, label="Matching-direction proportion")

    return plt

def plot_all_direction_profiles(
    profiles: np.ndarray,
    Az: np.ndarray,
    El: np.ndarray,
    title: str,
    cmap: str = "viridis",
    vmin: float = 0.0,
    vmax: float = 1.0,
):
    """Plot one profile matrix at every measured (Az, El) direction."""
    profiles = np.asarray(profiles)
    Az = np.asarray(Az)
    El = np.asarray(El)

    if profiles.ndim != 3:
        raise ValueError("profiles must have shape (directions, rows, columns).")
    if profiles.shape[0] != Az.size or Az.shape != El.shape:
        raise ValueError("profiles, Az, and El must contain the same directions.")

    azimuths = np.sort(np.unique(Az))
    elevations = np.sort(np.unique(El))
    azimuth_index = {value: index for index, value in enumerate(azimuths)}
    elevation_index = {value: index for index, value in enumerate(elevations)}

    figure, axes = plt.subplots(
        len(elevations),
        len(azimuths),
        figsize=(24, 24),
        squeeze=False,
    )

    for axis in axes.flat:
        axis.axis("off")

    first_image = None
    for direction_index, (azimuth, elevation) in enumerate(zip(Az, El)):
        # Reverse rows so positive elevation appears toward the top of the figure.
        row = len(elevations) - 1 - elevation_index[elevation]
        column = azimuth_index[azimuth]
        axis = axes[row, column]

        first_image = axis.imshow(
            profiles[direction_index],
            aspect="auto",
            origin="lower",
            cmap=cmap,
            vmin=vmin,
            vmax=vmax,
        )
        axis.set_title(
            f"Az={azimuth:g}, El={elevation:g}",
            fontsize=4,
            pad=1,
        )
        axis.set_xticks([])
        axis.set_yticks([])
        axis.axis("on")

    figure.suptitle(f"{title} ({profiles.shape[0]} directions)", fontsize=16)
    figure.subplots_adjust(
        left=0.01,
        right=0.97,
        bottom=0.01,
        top=0.94,
        wspace=0.05,
        hspace=0.35,
    )

    # Keep the shared colorbar compact and out of the diamond layout.
    colorbar_axis = figure.add_axes([0.76, 0.955, 0.20, 0.012])
    colorbar = figure.colorbar(first_image, cax=colorbar_axis, orientation="horizontal")
    colorbar.set_label("Matching-direction proportion", labelpad=2)

    return figure, axes
```

# Convert data into tensorflow dataset

```python
import os
import tensorflow as tf
from sklearn.model_selection import train_test_split
```

```python
def convert_data_to_np(
    IPD_data: np.ndarray,
    ILD_data: np.ndarray,
    Az: np.ndarray,
    El: np.ndarray,
    f: np.ndarray,
    test_size: float = 0.2,
    save_dir: str = "data/np_arrays",
    ):

    assert IPD_data.ndim == 3, "IPD data must have shape (directions, frequencies, target_axis_values)."
    assert ILD_data.ndim == 3, "ILD data must have shape (directions, frequencies, target_axis_values)."
    assert Az.ndim == 1, "Az must be a 1D array."
    assert El.ndim == 1, "El must be a 1D array."
    assert IPD_data.shape[0] == ILD_data.shape[0] == Az.size == El.size, "IPD, ILD, Az, and El must contain the same number of directions."
    assert IPD_data.shape[1] == ILD_data.shape[1], "IPD and ILD data must have the same number of frequencies."

    # Flatten each direction's IPD and ILD data into a single vector | shape: (directions, freq*n_ipd_bins) and (directions, freq*n_ild_bins)
    x_ipd = IPD_data.reshape(IPD_data.shape[0], -1).astype(np.float32)
    x_ild = ILD_data.reshape(ILD_data.shape[0], -1).astype(np.float32)

    # Concatenate IPD and ILD data along the last axis | shape: (directions, freq*n_ipd_bins + freq*n_ild_bins)
    x = np.concatenate([x_ipd, x_ild], axis=-1)

    # Convert Az and El to a single array of shape (directions, 2)
    y = np.stack([Az, El], axis=1).astype(np.float32)

    # Train / test split
    x_train, x_test, y_train, y_test = train_test_split(
        x, y, test_size=test_size, random_state=42
    )

    # Save the data
    np.savez_compressed(
        os.path.join(save_dir, "neurological_data.npz"),
        x=x,
        y=y,
        x_train=x_train,
        x_test=x_test,
        y_train=y_train,
        y_test=y_test,
        ITD_data=IPD_data,
        ILD_data=ILD_data,
        Az=Az,
        El=El,
        f=f
    )

    print(f"x shape:          {x.shape}")
    print(f"y shape:          {y.shape}")
    print(f"input_dim:        {x.shape[1]}")
    print(f"train samples:    {x_train.shape[0]}")
    print(f"test samples:     {x_test.shape[0]}")

    return x_train, x_test, y_train, y_test

```

```python
# Save the data to a directory
save_dir = r"/home/mrsarti/git_repos/rANN/data/neurological"

_, _, _, _ = convert_data_to_np(
    IPD_data=IPD_np,
    ILD_data=ILD_np,
    Az=Az,
    El=El,
    f=f,
    test_size=0.2,
    save_dir=save_dir
)
```
