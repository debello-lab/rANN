# Convert Acoustic Data to Neurological

```python
def average_over_target_axis(
        values: np.ndarray, 
        Az: np.ndarray, 
        El: np.ndarray, 
        target: str,
        verbose: int,
        ):
    """
    Averages the values over the specified target axis.

    Parameters:
    values (numpy.ndarray): The input array to be averaged.
    Az (numpy.ndarray): The azimuth values.
    El (numpy.ndarray): The elevation values.
    target (str): The target axis to average over.

    Returns:
    numpy.ndarray: The averaged array.
    """
    assert values.ndim == 2, "Input values must be a 2D array."
    assert Az.ndim == 1 and El.ndim == 1, "Az and El must be 1D arrays."
    assert target in ["Az", "El"], "Target must be either 'Az' or 'El'."


    if target == "Az":
        target_values, target_group = np.unique(Az, return_inverse=True)
    elif target == "El":
        target_values, target_group = np.unique(El, return_inverse=True)



    averaged = np.empty((values.shape[0], target_values.size))

    for group_idx in range(target_values.size):
        group_mask = target_group == group_idx
        averaged[:, group_idx] = np.mean(values[:, group_mask], axis=1)

    if verbose > 0:
        print(f"target_values_len: {len(target_values)}")
        print(f"target_group_min: {target_group.min()}, target_group_max: {target_group.max()}")
        print(f"target_group_len: {len(target_group)}")
        print(f"averaged shape: {averaged.shape}")  # should be (52, 37)


    return averaged
```

```python
def convert_acoustic_to_neurological(
        IXD: np.ndarray, 
        Az: np.ndarray, 
        El: np.ndarray, 
        f: np.ndarray, 
        target: str = "Az",
        tol: float = 1.5,
        verbose: int = 0,
        plot_segments: bool = False
        ):

    assert target in ["Az", "El"], "Target must be either 'Az' or 'El'"

    IXD_averaged_neurological_profile = []
        
    for idx, (az, el) in enumerate(zip(Az, El)):
                
        print(f"Index: {idx}, Azimuth: {az}, Elevation: {el}")

        direction_index = np.where((Az == az) & (El == el))

        if verbose > 5:
            print(f"Direction Index: {direction_index}")

        # Find az, el combination in data set 
        ixd = np.squeeze(IXD[direction_index])

        if verbose > 5:
            print(f"ixd shape: {ixd.shape}") # should be (52,)
            print(f"Index: {idx}, Azimuth: {az}, Elevation: {el}")

        IXD_y = np.zeros((f.size, IXD.shape[0] ))  # Initialize an empty array to store the similar values

        for f_index, ixd_f in enumerate(ixd):
            if verbose > 5:
                print(f"Frequency: {f[f_index]}, Value: {ixd_f}") # Gives us the value at that freq index 

            ixd_y = IXD[:,f_index] # size (685,) 

            # Find similar values within the tolerance
            ixd_similar_indices = np.where(np.isclose(ixd_y, ixd_f, atol=tol))[0]

            if ixd_similar_indices.size > 0:
                if verbose > 5:
                    print(f"ixd_y_similar shape: {ixd_y[ixd_similar_indices].shape}") # should be (685,)
                IXD_y[f_index, ixd_similar_indices] = 1


            if plot_segments and f_index % 100 == 0:
                # Across all other az, el combinations at same f_index find within a tolerence values and denote their location

                if verbose > 5:
                    print(f"ixd_y shape: {ixd_y.shape}") # should be (685,)

                dir = np.linspace(-90,90,37)
                #print(f"dir shape: {dir.shape}") # should be (37,)
                Maz, Mel = np.meshgrid(dir,dir)
                #print(f"Maz shape: {Maz.shape}, Mel shape: {Mel.shape}") # should be (37, 37)

                indx, indy = np.where(np.abs(Maz) + np.abs(Mel) <=90)

                plt.subplot(1, 2, 1)

                plt.subplot(1,2,1)
                P = np.zeros((len(dir), len(dir)))*np.nan
                P[indy, indx] = ixd_y
                plt.contourf(dir, dir, P)
                plt.colorbar()
                plt.xlabel('Azimuth (deg)')
                plt.ylabel('Elevation (deg)')
                #plt.title(f"Frequency: {f[f_index]}, Value: {ixd_f}")

                plt.subplot(1,2,2)
                P2 = np.zeros((len(dir), len(dir)))*np.nan
                P2[indy[ixd_similar_indices], indx[ixd_similar_indices]] = ixd_y[ixd_similar_indices]
                plt.contourf(dir, dir, P2)
                plt.xlabel('Azimuth (deg)')
                plt.colorbar()
                plt.show()

        IXD_segment = average_over_target_axis(IXD_y, Az, El, target=target, verbose=verbose)

        if plot_segments and idx % 100 == 0: 
            plot_neurological_profile(IXD_segment, az, el, f[f_index], target=target)


        IXD_averaged_neurological_profile.append(IXD_segment)


    print(f"IXD_averaged_neurological_profile shape: {len(IXD_averaged_neurological_profile)}, {len(IXD_averaged_neurological_profile[0])}") # should be (52, 685)

    return IXD_averaged_neurological_profile
```

```python
def plot_neurological_profile(IXD_segment,az, el, f, target="Az", verbose=0):
    plt.figure(figsize=(8, 8))
    plt.title(f"Averaged Segment - Frequency: {f:.2f}, Az: {az}, El: {el}")
    if verbose > 0:
        print(f"Averaged Segment shape: {len(IXD_segment)}, {len(IXD_segment[0])}") # should be (52, 685)
    plt.ylabel('Frequency (kHz)')
    plt.xlabel(f'Target Axis: {target}')
    plt.imshow(IXD_segment, cmap='viridis')
    plt.show()
    
    return plt
```
