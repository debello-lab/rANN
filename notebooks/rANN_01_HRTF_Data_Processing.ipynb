```python
# Imports
import matplotlib.pyplot as plt
import numpy as np
import scipy.io as sio
```

# Load & Plot HRTF Data

```python
#Load HRTF data
!wget https://www.dropbox.com/s/tvbv9hxwu2ydtbz/HRTF880ILDIPDABL.mat?dl=0

#Filename
filename = 'HRTF880ILDIPDABL.mat?dl=0'

#The data are .mat files. Use scipy to load a file:
mat = sio.loadmat(filename, squeeze_me=True)
```

```python
# Visualize all data
mat
```

```python
#HRTF properties
IPD = mat['IPD']
ILD = mat['ILD']
ABL = mat['ABL']

#Directions
Az = mat['Xc']
El = mat['Yc']

#Frequency
f = mat['f2']
```

```python
# Example IPD 
print(f"IPD values: {IPD}")
print(f"IPD min: {IPD.min()}, max: {IPD.max()}")

# Example ILD
print(f"ILD values: {ILD}")
print(f"ILD min: {ILD.min()}, max: {ILD.max()}")
```

```python
# Validate size of arrays
print(f'Frequency Shape: {f.shape}')
print(f'IPD Shape: {IPD.shape}')
print(f'ILD Shape: {ILD.shape}')
print(f'ABL Shape: {ABL.shape}')
print(f'Azimuth Shape: {Az.shape}')
print(f'Elevation Shape: {El.shape}')
```

```python
# Plot Azimuth vs Elevation span
plt.scatter(Az,El)
plt.xlabel('Azimuth (deg)', fontsize = 14)
plt.ylabel('Elevation (deg)', fontsize = 14)
plt.show()
```

```python
#Plot IPD and ILD for selected directions
count = 5
step = len(list(zip(Az, El))) // count

for idx, (az, el) in enumerate(zip(Az, El)):
    if idx % step == 0:
        print(f"Index: {idx}, Azimuth: {az}, Elevation: {el}")
        #Find the index of this direction in the direction matrices

        direction_index = np.where((Az == az) & (El == el))[0][0]

        #Plot
        plt.subplots(1,2, figsize = (13,5))

        plt.subplot(1,2,1)
        plt.plot(np.unwrap(IPD[direction_index,]), f)
        plt.xlim(-np.pi, np.pi)
        plt.ylabel('Frequency (kHz)', fontsize = 14)
        plt.xlabel('IPD (radians)', fontsize = 14)

        plt.subplot(1,2,2)
        plt.plot(ILD[direction_index,], f)
        plt.xlim(-30, 30)
        plt.ylabel('Frequency (kHz)', fontsize = 14)
        plt.xlabel('ILD (dB)', fontsize = 14)
        plt.show()

```

# Convert Acoustic Data to Neurological

```python
IXD = IPD
tol = 0.5  # Tolerance for finding similar values

for idx, (az, el) in enumerate(zip(Az, El)):
    print(f"Index: {idx}, Azimuth: {az}, Elevation: {el}")


    direction_index = np.where((Az == az) & (El == el))
    print(f"Direction Index: {direction_index}")


    # Find az, el combination in data set 
    ixd = np.squeeze(IXD[direction_index])

    print(f"ixd shape: {ixd.shape}") # should be (52,)

    print(f"Index: {idx}, Azimuth: {az}, Elevation: {el}")

    for f_index, ixd_f in enumerate(ixd):
        print(f"Frequency Index: {f_index}, Value: {ixd_f}") # Gives us the value at that freq index 


        # Across all other az, el combinations at same f_index find within a tolerence values and denote their location
        ixd_y = IXD[:,f_index] # size (685,) 

        print(f"ixd_y shape: {ixd_y.shape}") # should be (685,)

        dir = np.linspace(-90,90,37)
        Maz, Mel = np.meshgrid(dir,dir)

        indx, indy = np.where(np.abs(Maz) + np.abs(Mel) <=90)
        
        P = np.zeros((len(dir), len(dir)))*np.nan
        P[indy, indx] = ixd_y
        plt.contourf(dir, dir, P)
        plt.colorbar()
        plt.title(f"Frequency Index: {f_index}, Value: {ixd_f}")
        plt.show()

        # Find similar values within the tolerance
        ixd_similar_indices = np.where(np.isclose(ixd_y, ixd_f, atol=tol))[0]
        print(f"Similar indices for frequency index {f_index}: {ixd_similar_indices}")


        P = np.full((len(dir), len(dir)), np.nan)
        P[indy[ixd_similar_indices], indx[ixd_similar_indices]] = ixd_y[ixd_similar_indices]

        plt.contourf(dir, dir, P)
        plt.xlabel('Azimuth (deg)', fontsize = 14)
        plt.ylabel('Elevation (deg)', fontsize = 14)
        plt.title(f"Frequency Index: {f_index}, Value: {ixd_f}")
        plt.colorbar()
        plt.show()

        break

    break
```

```python
#Create data set of x, y for each elevation  
x_data = []
y_data = []
freq_axis = f
itd_axis = IPD[0]
ild_axis = ILD[0]

for idx, (az, el) in enumerate(zip(Az, El)):
    direction_index = np.where((Az == az) & (El == el))[0][0]

    if IPD[direction_index,].size == 0 or ILD[direction_index,].size == 0:
        print(idx)
        continue

    # print('Shape of f:', f.shape)
    # print('Shape of IPD:', IPD[direction_index,].shape)
    # print('Shape of ILD:', ILD[direction_index,].shape)

    features = np.concatenate(
                [
                    #np.stack(f, axis=0).reshape(-1),
                    np.stack(IPD[direction_index,], axis=0).reshape(-1),
                    np.stack(ILD[direction_index,], axis=0).reshape(-1),
                ]
            ).astype(np.float32)

    # print('Shape of data:', features.shape)

    x_data.append(features)
    y_data.append([az, el])

x_data = np.array(x_data)
y_data = np.array(y_data)
```
